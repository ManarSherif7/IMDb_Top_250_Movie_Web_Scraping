```python
from selenium import webdriver
```

```python
driver= webdriver.chrome()
```

```python
driver.get("https://www.imdb.com/chart/top/?ref_=hm_nv_menu")
```

```python
moves_names= driver.find_elements("xpath","//h4[@class='ipc-title__text']")

```

```python
for names in moves_names:
    print(names.text)
moves_names=moves_names[:250]
```

```python
len(moves_names)

```

```python
move_ratings= driver.find_elements("xpath","//li[contains(@class,'ipc-metadata-list-summary-item')]//span[contains(@class, 'ipc-rating-star--rating')]")


```

```python
for ratings in move_ratings:
    print(ratings.text)
```

```python
move_date= driver.find_elements("xpath","//li[contains(@class, 'ipc-metadata-list-summary-item')]//div[contains(@class, 'cli-title-metadata')]//li[contains(@class, 'ipc-inline-list__item')][1]")


```

```python
for date in move_date:
    print(date.text)
```

```python
move_duration=driver.find_elements("xpath","//li[contains(@class, 'ipc-metadata-list-summary-item')]//div[contains(@class, 'cli-title-metadata')]//li[2]")
```

```python
for duration in move_duration:
    print(duration.text)
```

```python
import pandas as pd
```

```python


df=pd.DataFrame({"Movies Names": [i.text for i in moves_names],"Ratings":[i.text for i in move_ratings],"Produce Year":[i.text for i in move_date],"Duration":[i.text for i in move_duration]})
```

```python
print(df)

```

```python
df.index=range(1,len(df)+1)
df.index.name='Movie Rank'
df=df.reset_index()
print(df)
```

```python
import os
desktop_path = os.path.join(os.path.expanduser("~"), "Desktop")
file_path = os.path.join(desktop_path, "IMDb_Top_250_Movies.xlsx")
df.to_excel(file_path, index=False)
```
