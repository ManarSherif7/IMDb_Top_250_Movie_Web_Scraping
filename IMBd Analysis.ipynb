```python
import pandas as pd
import seaborn as sn
import os
import matplotlib.pyplot as plt
```

```python

desktop_path = os.path.join(os.path.expanduser("~"), "Desktop")
file_path = os.path.join(desktop_path, "IMDb_Top_250_Movies.xlsx")

df = pd.read_excel(file_path)

print("Data loaded successfully! Rows & Columns:", df.shape)
print(df.head())


```

```python

plt.figure(figsize=(9, 5))
sn.histplot(df['Ratings'], bins=10, kde=True, color='purple')
plt.title('Distribution of Ratings in IMDb Top 250', fontsize=14, fontweight='bold')
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

```

```python
plt.savefig('ratings_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

```

```python
plt.figure(figsize=(12, 6))
top_years = df['Produce Year'].value_counts().head(10)

sn.barplot(
x=top_years.index, 
y=top_years.values, 
hue=top_years.index, 
palette='viridis', 
legend=False
)

plt.title('Top 10 Years with Most Movies in IMDb Top 250', fontsize=14, fontweight='bold')
plt.xlabel('Production Year', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
```

```python
plt.savefig('top_years.png', dpi=300, bbox_inches='tight')
plt.show()
```
